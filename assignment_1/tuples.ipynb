```python
# Create the tuple
ages = (20, 25, 22, 30, 28)

# Print the tuple
print(ages)
```

```python
# First element
print( "first element:",ages[0])

# Last element
print( "last element:",ages[-1])
```

```python
# Maximum value
print("maximum value:", max(ages))

# Minimum value
print("minimum value:", min(ages) )

# Number of elements
print("number of elements:", len(ages))
```
