```python
# 1. Create a dictionary containing the population of some cities

population = {
    "Lima": 10092000,
    "Arequipa": 1008000,
    "Trujillo": 919000
}
```

```python
# 2. Print the complete dictionary

print(population)
```

```python
# 3.Display only the keys of the dictionary

keys = population.keys()
print(keys)
```

```python
# 4. Get the population of "Lima"

Lima_population = population.get("Lima")
print("Population of Lima:", Lima_population)
```

```python
# 5. Add "Cuscio" to the dictionary with a population of 428000

population["Cusco"] = 428000
print(population)
```

```python
# 6. Remove "Trujillo"

population.pop("Trujillo", None)
print(population)
```

```python
# 7. Print final dictionary

print("Final dictionary:", population)
```
